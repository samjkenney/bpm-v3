```python
# plot phase % error 

import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

df = pd.read_csv('../grace-7222025/cleaned.csv')
df['percent-error'] = df['percent-error']*100

phase_1_df = df[df['Height'] == 7]
phase_2_df = df[df['Height'] == 8]

df = pd.read_csv('../grace-7232025/cleaned.csv')
df['percent-error'] = df['percent-error']*100

phase_3_df = df[df['Height'] == 1]
phase_4_df = df[df['Height'] == 2]
phase_5_df = df[df['Height'] == 3]

plt.plot(np.array(phase_1_df['Time'].to_list()), 
         np.array(phase_1_df['deviation'].to_list()), label = 'Profile 1')


plt.plot(np.array(phase_2_df['Time'].to_list()), 
         np.array(phase_2_df['deviation'].to_list()), label = 'Profile 2')

plt.plot(np.array(phase_3_df['Time'].to_list()), 
         np.array(phase_3_df['deviation'].to_list()), label = 'Profile 3')

plt.plot(np.array(phase_4_df['Time'].to_list()), 
         np.array(phase_4_df['deviation'].to_list()), label = 'Profile 4')

plt.plot(np.array(phase_5_df['Time'].to_list()), 
         np.array(phase_5_df['deviation'].to_list()), label = 'Profile 5')

plt.xlabel("Time (s)")
plt.ylabel('Percent Error')
plt.legend()
plt.minorticks_on()
plt.grid(which='major', linestyle='-') # Major gridlines
plt.grid(which='minor', linestyle=':') # Minor gridlines
plt.title('Real Error vs Time Across Radar Profiles')

plt.savefig('./profile-accuracy.png')
```

```python
# plot num_distances_to_analyze 
from matplotlib.lines import Line2D

df = pd.read_csv('../grace-7222025/cleaned.csv')
df['percent-error'] = df['percent-error']*100

filler_lines = [Line2D([0],[0],color='blue'), Line2D([0],[0],color='red')]

dist_3_df = df[df['Height'] == 1]
plt.plot(np.array(dist_3_df['Time'].to_list()),np.array(dist_3_df['percent-error'].to_list()), color='blue', linestyle = '-')

dist_3_df = df[df['Height'] == 2]
plt.plot(np.array(dist_3_df['Time'].to_list()),np.array(dist_3_df['percent-error'].to_list()), color='blue', linestyle = '--')

dist_3_df = df[df['Height'] == 3]
plt.plot(np.array(dist_3_df['Time'].to_list()),np.array(dist_3_df['percent-error'].to_list()), color='blue', linestyle = 'dotted')


dist_3_df = df[df['Height'] == 4]
plt.plot(np.array(dist_3_df['Time'].to_list()),np.array(dist_3_df['percent-error'].to_list()), color='red', linestyle = '-')

dist_3_df = df[df['Height'] == 5]
plt.plot(np.array(dist_3_df['Time'].to_list()),np.array(dist_3_df['percent-error'].to_list()), color='red', linestyle = '--')

dist_3_df = df[df['Height'] == 6]
plt.plot(np.array(dist_3_df['Time'].to_list()),np.array(dist_3_df['percent-error'].to_list()), color='red', linestyle = 'dotted')

plt.legend(filler_lines, ['3 Distances','5 Distances'],loc = 'upper right')
plt.show()

```

```python
# plot sweeps per frame
df = pd.read_csv('../grace-7292025/cleaned.csv')

sweeps_32 = df[df["Height"]==1]
sweeps_16 = df[df["Height"]==2]
sweeps_8 = df[df["Height"]==3]


plt.plot(np.array(sweeps_32["Time"].to_list()), np.array(sweeps_32["deviation"].to_list()), label = "32 SPF")
plt.plot(np.array(sweeps_16["Time"].to_list()), np.array(sweeps_16["deviation"].to_list()), label = "16 SPF")
plt.plot(np.array(sweeps_8["Time"].to_list()), np.array(sweeps_8["deviation"].to_list()), label = "8 SPF")

plt.xlabel("Time (s)")
plt.ylabel('Real Error')
plt.legend()
plt.minorticks_on()
plt.grid(which='major', linestyle='-') # Major gridlines
plt.grid(which='minor', linestyle=':') # Minor gridlines
plt.title('Real Error vs Time Across Sweeps per Frame (SPF)')

plt.savefig('./spf-accuracy.png')
```

```python
# 6,9,12 bpm visualization
df = pd.read_csv('../final-trials/cleaned.csv')

bpm_6_df = df[df["Height"]==1]
bpm_9_df = df[df["Height"]==2]
bpm_12_df = df[df["Height"]==3]

default_6 = df[df["Height"]==4]
default_9 = df[df["Height"]==5]
default_12 = df[df["Height"]==6]

sam_6 = df[df["Height"]==7]
sam_9 = df[df["Height"]==8]
sam_12 = df[df["Height"]==9]
# set up partitioned plot
fig,(ax1,ax2,ax3) = plt.subplots(1,3,sharey=True,sharex=True)

ax1.plot(np.array(bpm_6_df["Time"].to_list()), np.array(bpm_6_df["deviation"].to_list()))
ax1.plot(np.array(default_6["Time"].to_list()), np.array(default_6["deviation"].to_list()))
ax1.plot(np.array(sam_6["Time"].to_list()), np.array(sam_6["deviation"].to_list()))
ax1.set_title("6 BPM")
ax2.plot(np.array(bpm_9_df["Time"].to_list()), np.array(bpm_9_df["deviation"].to_list()))
ax2.plot(np.array(default_9["Time"].to_list()), np.array(default_9["deviation"].to_list()))
ax2.plot(np.array(sam_9["Time"].to_list()), np.array(sam_9["deviation"].to_list()))
ax2.set_title("9 BPM")
ax3.plot(np.array(bpm_12_df["Time"].to_list()), np.array(bpm_12_df["deviation"].to_list()))
ax3.plot(np.array(default_12["Time"].to_list()), np.array(default_12["deviation"].to_list()))
ax3.plot(np.array(sam_12["Time"].to_list()), np.array(sam_12["deviation"].to_list()))
ax3.set_title("12 BPM")

fig.subplots_adjust(wspace=0)
fig.suptitle("Deviation vs Time Across Tempoed Breath Rates")

plt.show()
```
